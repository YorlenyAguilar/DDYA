## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscrip(n,nemo):
    if n==0:
        return 0
    if n==1:
        return 1
    if n in nemo:
        return nemo [n]
    nemo[n] = suscrip(n-1, nemo) + suscrip(n-2, nemo)

    return nemo[n]
```
