### 5) Análisis de complejidad **[15 pts]**
Incluye:

**Justificación**

Sea n el número de vértices y E el número de aristas. En el grafo completo, E = n(n-1)/2.

Calcular distancias de todas las parejas: O(E) = O(n²).
Ordenar aristas: O(E log E).
Inicializar Union-Find: O(n).
Recorrer aristas con find/union: O(E · α(n)), con α(n) casi constante.

**Función T[n]**

T(n) = c1·n·log2(n) + c2·E·α(n) + c3·n, con E = n(n-1)/2. Para este laboratorio, n = 7, E = 21 y E·log2(E) ≈ 92 operaciones de ordenamiento.

**Mejor caso**

Las n-1 aristas más livianas ya forman un árbol (por ejemplo, componentes alineados). Kruskal acepta n-1 aristas y se detiene, examinando solo n-1 después de ordenar. Costo total: O(E log E) por el ordenamiento.

**peor caso**

La última arista necesaria es la más pesada (por ejemplo, dos grupos unidos por un único puente pesado). Se examinan las E aristas. Costo total: O(E log E).

```python

import matplotlib.pyplot as plt
filas = "ABCDEFGHIJ"
posiciones_originales = {
    "F": "A3",    # Fuente
    "S": "B16",    # Sensor
    "L": "C8",    # LED
    "P": "D24",    # Pulsador
    "M": "G24",    # Microcontrolador
    "B": "H6",    # Buzzer
    "R": "I13",    # Potenciómetro
}

nombres = {"F": "Fuente", "S": "Sensor", "L": "LED", "P": "Pulsador",
           "M": "Microcontrolador", "B": "Buzzer", "R": "Potenciómetro"}

colores = {"F": "red", "S": "green", "L": "orange", "P": "blue",
           "M": "purple", "B": "brown", "R": "gray"}
def a_coordenadas(pos):
    """Convierte "H6" en (columna, fila) = (6, 8). La letra se vuelve número (A=1 ... J=10)."""
    columna = int(pos[1:])
    fila = filas.index(pos[0]) + 1
    return (columna, fila)


def calcular_coordenadas(posiciones):
    """Convierte todo el diccionario {"F": "A3", ...} en {"F": (3, 1), ...}."""
    coords = {}
    for comp in posiciones:
        coords[comp] = a_coordenadas(posiciones[comp])
    return coords


def distancia(p1, p2):
    """Distancia Manhattan: |columna1 - columna2| + |fila1 - fila2|. Es el peso de la arista."""
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])


def construir_aristas(coords):
    """
    Construye el grafo: une cada componente con todos los demás.
    Cada arista es (peso, componente1, componente2); el peso se CALCULA
    con distancia(). Con 7 componentes salen 21 aristas.
    """
    aristas = []
    comps = list(coords)
    for i in range(len(comps)):
        for j in range(i + 1, len(comps)):
            a = comps[i]
            b = comps[j]
            aristas.append((distancia(coords[a], coords[b]), a, b))
    return aristas


def buscar(padre, x):
    """Union-Find: devuelve el 'jefe' del grupo de x. Mismo jefe = ya conectados (ciclo)."""
    while padre[x] != x:
        x = padre[x]
    return x


def kruskal(coords, aristas, mostrar=True):
    """
    Kruskal: ordena aristas de menor a mayor peso, acepta la arista si une
    dos grupos distintos (sin ciclo) y se detiene al tener n-1 aristas.
    Si mostrar=True imprime el paso a paso.
    Devuelve la lista de cables elegidos (mst) y la longitud total.
    """
    padre = {c: c for c in coords}
    mst = []
    total = 0
    for peso, a, b in sorted(aristas):
        raiz_a = buscar(padre, a)
        raiz_b = buscar(padre, b)
        if raiz_a != raiz_b:
            padre[raiz_b] = raiz_a
            mst.append((peso, a, b))
            total = total + peso
            if mostrar:
                print(f"  ✔ {a}-{b} ({peso}) aceptada")
        elif mostrar:
            print(f"  ✘ {a}-{b} ({peso}) rechazada: forma ciclo")
        if len(mst) == len(coords) - 1:
            break
    return mst, total


def mostrar_resultado(mst, total, titulo):
    """Imprime cada cable elegido y la longitud total."""
    print(f"\n========== {titulo} ==========")
    for i in range(len(mst)):
        peso, a, b = mst[i]
        print(f"Cable {i + 1}: {a} - {b} = {peso}")
    print(f"Longitud total de cable: {total} unidades")


def dibujar(ax, coords, aristas, titulo, es_mst=False, resaltar=None):
    """
    Dibuja la protoboard (30 columnas x 10 filas) con los componentes y,
    si hay aristas, los cables enrutados de forma ORTOGONAL (punto a punto tipo L).
    """
    for c in range(1, 31):
        for f in range(1, 11):
            ax.plot(c, f, ".", color="lightgray")

    for peso, a, b in aristas:
        x1, y1 = coords[a]
        x2, y2 = coords[b]
        px = [x1, x2, x2]
        py = [y1, y1, y2]

        if es_mst:
            ax.plot(px, py, color="red", linewidth=3, solid_capstyle='round', solid_joinstyle='round', zorder=2)
        else:
            ax.plot(px, py, "--", color="steelblue", linewidth=1, zorder=2)
        ax.text(x2, y1, str(peso), fontsize=8,
                ha="center", va="center", backgroundcolor="white", zorder=5)

    for comp in coords:
        c, f = coords[comp]
        borde = "gold" if comp == resaltar else "black"
        ancho = 4 if comp == resaltar else 1
        ax.scatter(c, f, s=600, color=colores[comp], edgecolor=borde,
                   linewidth=ancho, zorder=3)
        ax.text(c, f, comp, color="white", ha="center", va="center",
                fontweight="bold", zorder=4)
        ax.text(c, f + 0.6, nombres[comp] + " " + filas[f - 1] + str(c),
                ha="center", fontsize=7)

    ax.set_xticks(range(1, 31))
    ax.set_yticks(range(1, 11))
    ax.set_yticklabels(list(filas))
    ax.set_ylim(11, 0)
    ax.xaxis.tick_top()
    ax.set_xlabel("Columnas")
    ax.set_ylabel("Filas")
    ax.set_title(titulo, fontweight="bold")
    ax.set_aspect("equal")
def opcion_ver_posiciones(posiciones):
    """Opción 1: muestra la posición original y la actual de cada componente."""
    print("\nPOSICIONES DE LOS COMPONENTES")
    print(f"{'Componente':<24}{'Original':<10}{'Actual':<10}")
    for comp in posiciones_originales:
        marca = "  <- movido" if posiciones[comp] != posiciones_originales[comp] else ""
        print(f"{comp + ' ' + nombres[comp]:<24}{posiciones_originales[comp]:<10}"
              f"{posiciones[comp]:<10}{marca}")


def opcion_ver_protoboard(posiciones):
    """Opción 2: dibuja la protoboard con las posiciones actuales."""
    coords = calcular_coordenadas(posiciones)
    fig, ax = plt.subplots(figsize=(16, 7))
    dibujar(ax, coords, [], "Protoboard con las posiciones actuales")
    plt.show()


def opcion_ver_conexiones(posiciones):
    """Opción 3: lista y dibuja TODAS las posibles conexiones con su peso."""
    coords = calcular_coordenadas(posiciones)
    aristas = construir_aristas(coords)
    print(f"\nTODAS LAS POSIBLES CONEXIONES ({len(aristas)} aristas)")
    for peso, a, b in sorted(aristas):
        print(f"{a} -------- {peso} -------- {b}")
    fig, ax = plt.subplots(figsize=(16, 7))
    dibujar(ax, coords, aristas, "Todas las posibles conexiones (peso = distancia)")
    plt.show()


def opcion_kruskal(posiciones):
    """Opción 4: ejecuta Kruskal con las posiciones actuales y dibuja el MST."""
    coords = calcular_coordenadas(posiciones)
    aristas = construir_aristas(coords)
    print("\nKruskal paso a paso:")
    mst, total = kruskal(coords, aristas)
    mostrar_resultado(mst, total, "RESULTADO")
    fig, ax = plt.subplots(figsize=(16, 7))
    dibujar(ax, coords, mst, f"Conexiones mínimas (Kruskal) - Total: {total}", es_mst=True)
    plt.show()


def opcion_mover_buzzer(posiciones):
    """
    Opción 5: mueve el Buzzer a J20. Las posiciones originales se conservan
    aparte (posiciones_originales) para poder comparar después.
    """
    if posiciones["B"] == "J20":
        print("\nEl Buzzer ya está en J20.")
        return
    anterior = posiciones["B"]
    posiciones["B"] = "J20"
    print(f"\nBuzzer movido: {anterior} -> J20")
    print("El grafo y el MST se recalcularán automáticamente (opciones 4 y 6).")


def opcion_comparar(posiciones):
    """
    Opción 6: compara el MST original con el MST actual y dibuja ambas protoboards.
    """
    if posiciones["B"] == posiciones_originales["B"]:
        print("\nPrimero mueve el Buzzer con la opción 5.")
        return

    # MST original
    coords1 = calcular_coordenadas(posiciones_originales)
    mst1, total1 = kruskal(coords1, construir_aristas(coords1), mostrar=False)
    # MST después del cambio
    coords2 = calcular_coordenadas(posiciones)
    mst2, total2 = kruskal(coords2, construir_aristas(coords2), mostrar=False)

    mostrar_resultado(mst1, total1, "MST ORIGINAL")
    mostrar_resultado(mst2, total2, "MST DESPUÉS DEL CAMBIO")

    print("\n==================== COMPARACIÓN ====================")
    print(f"MST original           Longitud total: {total1}")
    print(f"MST después del cambio  Longitud total: {total2}")
    if total2 > total1:
        print(f"¿Aumentó o disminuyó? AUMENTÓ en {total2 - total1} unidades de cable.")
    elif total2 < total1:
        print(f"¿Aumentó o disminuyó? DISMINUYÓ en {total1 - total2} unidades de cable.")
    else:
        print("¿Aumentó o disminuyó? NO CAMBIÓ: la cantidad total de cable es la misma.")

    c1 = {(a, b) for p, a, b in mst1}
    c2 = {(a, b) for p, a, b in mst2}
    print("Cables que ya no se usan:", [a + "-" + b for a, b in c1 - c2] or "ninguno")
    print("Cables nuevos:            ", [a + "-" + b for a, b in c2 - c1] or "ninguno")

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 14))
    dibujar(ax1, coords1, mst1,
            f"MST original (Buzzer en {posiciones_originales['B']}) - Total: {total1}", es_mst=True)
    dibujar(ax2, coords2, mst2,
            f"MST después del cambio (Buzzer en {posiciones['B']}) - Total: {total2}",
            es_mst=True, resaltar="B")
    plt.show()

    fig, ax = plt.subplots(figsize=(6, 4))
    barras = ax.bar(["Original", "Después del cambio"], [total1, total2],
                    color=["steelblue", "purple"])
    ax.bar_label(barras, fontweight="bold")
    ax.set_ylabel("Longitud total de cable")
    ax.set_title("Comparación de cable necesario")
    ax.set_ylim(0, max(total1, total2) * 1.2)
    plt.show()


def opcion_restablecer(posiciones):
    """Opción 7: vuelve todos los componentes a sus posiciones originales."""
    posiciones.clear()
    posiciones.update(posiciones_originales)
    print("\nPosiciones restablecidas (Buzzer en H6).")
def menu():
    posiciones = dict(posiciones_originales)    # copia de trabajo
    while True:
        print("\n" + "=" * 52)
        print("    PROTOBOARD - ÁRBOL DE EXPANSIÓN MÍNIMA (KRUSKAL)")
        print("=" * 52)
        print("1. Ver posiciones de los componentes")
        print("2. Ver la protoboard (gráfica)")
        print("3. Ver todas las posibles conexiones (grafo)")
        print("4. Ejecutar Kruskal con las posiciones actuales")
        print("5. Mover el Buzzer de H6 a J20  (Punto 6)")
        print("6. Comparar MST original vs MST después del cambio")
        print("7. Restablecer posiciones originales")
        print("0. Salir")
        opcion = input("Elige una opción: ").strip()

        if opcion == "1":
            opcion_ver_posiciones(posiciones)
        elif opcion == "2":
            opcion_ver_protoboard(posiciones)
        elif opcion == "3":
            opcion_ver_conexiones(posiciones)
        elif opcion == "4":
            opcion_kruskal(posiciones)
        elif opcion == "5":
            opcion_mover_buzzer(posiciones)
        elif opcion == "6":
            opcion_comparar(posiciones)
        elif opcion == "7":
            opcion_restablecer(posiciones)
        elif opcion == "0":
            print("Fin del programa.")
            break
        else:
            print("Opción no válida. Escribe un número del 0 al 7.")


menu()
```

```python

def esta_conectado(comps, arbol):
    """
    Verifica que con las aristas del árbol se pueda llegar de un componente
    a todos los demás (que no quede ningún componente suelto).
    Empieza con el primer componente y va sumando los que se conectan a él.
    """
    conectados = {comps[0]}
    cambio = True
    while cambio:
        cambio = False
        for p, a, b in arbol:
            if (a in conectados) != (b in conectados):   # une uno conectado con uno que no
                conectados.add(a)
                conectados.add(b)
                cambio = True
    return len(conectados) == len(comps)


print("========== TESTS ==========")
assert a_coordenadas("A3") == (3, 1)
assert a_coordenadas("J20") == (20, 10)
print("✔ Test 1: conversión de posiciones")


assert distancia(a_coordenadas("A3"), a_coordenadas("C8")) == 7
print("✔ Test 2: distancia F-L = 7 (ejemplo del enunciado)")
assert len(aristas) == 21
print("✔ Test 3: el grafo tiene 21 aristas")
assert len(mst) == 6 and len(mst2) == 6
print("✔ Test 4: los dos MST tienen 6 aristas")
assert total == sum(p for p, a, b in mst)
assert total2 == sum(p for p, a, b in mst2)
print("✔ Test 5: la longitud total es la suma de los cables")
assert esta_conectado(list(coords), mst)
assert esta_conectado(list(coords2), mst2)
print("✔ Test 6: todos los componentes quedan conectados")

pesos_ordenados = sorted(p for p, a, b in aristas)
assert total >= sum(pesos_ordenados[:6])
print("✔ Test 7: el total no es menor que la suma de las 6 aristas más cortas")
assert coords["B"] == (6, 8) and coords2["B"] == (20, 10)
print("✔ Test 8: el Buzzer cambió de H6 a J20")

print("\nTODOS LOS TESTS PASARON")
```
